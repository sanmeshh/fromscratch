# Bag of Words

Why bag of words first? Kn bas naam se easy lagta hai isiliye.


Bag of words is a preprocessing technique which converts the occurences of the frequently occuring words into a vector. This helps to catch the semantic meaning between two sentences through the vectors.It is called bag of words because any information about the document's order or structure are discarded. The model is only concerned whether the known words are there in the given Document or not.


image.png



```python
data= ["doubt thou the stars are fire; ",
       "doubt that the sun doth move",
       "doubt truth to be a liar",
       "but never doubt i love"
]
def cvfit(data):
  uniques=set()
  for sentence in data:
    for word in sentence.split():
      if len(word)>1:
        uniques.add(word)

  vocab={}
  for i,word in enumerate(sorted(uniques)):
    vocab[word]=i

  return vocab
```

```python
def cvtransform(data):
  matrix=[]
  vocab=cvfit(data)
  for sentence in data:
    sentence=[word for word in sentence.split() if len(word)>1]
    arr=np.zeros(len(vocab))

    for word in sentence:
      arr[vocab[word]]+=1
    arr=list(arr)

    matrix.append(arr)

  matrix=np.array(matrix)
  matrix=matrix.astype(int)

  return matrix









```

```python
cvfit(data)
```

```python
my_vector=(cvtransform(data))
```
